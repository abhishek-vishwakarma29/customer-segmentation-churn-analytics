```python
# High-value customers
high_value_customers = df_clean[
    df_clean["BalanceSegment"] == "High-balance"
]

# High-value churners
high_value_churners = high_value_customers[
    high_value_customers["Exited"] == 1
]

# Calculate high-value churn ratio
high_value_churn_ratio = (
    len(high_value_churners) / len(high_value_customers)
) * 100

# Calculate total balance associated with high-value churners
high_value_balance_at_risk = high_value_churners["Balance"].sum()

# Calculate total balance of all churned customers
total_churned_balance = df_clean.loc[
    df_clean["Exited"] == 1, "Balance"
].sum()

# Percentage of churned balance associated with high-value churners
high_value_balance_share = (
    high_value_balance_at_risk / total_churned_balance
) * 100

print("High-Value Customers:", len(high_value_customers))
print("High-Value Churners:", len(high_value_churners))
print("High-Value Churn Ratio:", round(high_value_churn_ratio, 2), "%")
print("Balance Associated with High-Value Churners:",
      round(high_value_balance_at_risk, 2))
print("High-Value Share of Total Churned Balance:",
      round(high_value_balance_share, 2), "%")
```

```python
# Overall churn rate
overall_churn_rate = df_clean["Exited"].mean() * 100

# High-value churn ratio
high_value_churn_ratio = (
    high_value_customers["Exited"].mean() * 100
)

# Geographic Risk Index
geographic_risk_index = (
    geography_summary["Churn_Rate"].max()
    / overall_churn_rate
)

# Engagement Drop Indicator
inactive_churn_rate = activity_churn_summary.loc[
    "Inactive", "Churn_Rate"
]

active_churn_rate = activity_churn_summary.loc[
    "Active", "Churn_Rate"
]

engagement_drop_indicator = (
    inactive_churn_rate - active_churn_rate
)

print("Overall Churn Rate:", round(overall_churn_rate, 2), "%")
print("High-Value Churn Ratio:", round(high_value_churn_ratio, 2), "%")
print("Geographic Risk Index:", round(geographic_risk_index, 2))
print("Engagement Drop Indicator:", round(engagement_drop_indicator, 2), "percentage points")
```
